```python
# ============================================
# GENERATIVE TEXT MODEL USING GPT-2
# ============================================
# Task:
# Generate coherent paragraphs based on user prompts
#
# Install Required Libraries:
# pip install transformers torch
# ============================================

from transformers import pipeline

# --------------------------------------------
# Load Pretrained GPT-2 Model
# --------------------------------------------
print("Loading GPT-2 Model...")

generator = pipeline(
    "text-generation",
    model="gpt2"
)

print("Model Loaded Successfully!")

# --------------------------------------------
# Function for Text Generation
# --------------------------------------------
def generate_text(prompt):

    result = generator(
        prompt,
        max_length=150,
        num_return_sequences=1,
        temperature=0.8
    )

    generated_text = result[0]['generated_text']

    return generated_text

# --------------------------------------------
# Main Program (Interactive loop removed)
# --------------------------------------------
# The interactive loop has been removed to allow for automated execution.
# A separate demonstration cell will be used to showcase text generation.
```
